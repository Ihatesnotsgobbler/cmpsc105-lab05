## Plot 1

**Scenario**: A researcher wants to show that Gentoo penguins are significantly heavier than other species, but their plot is visually overwhelming and exaggerates the differences.

```python
import seaborn as sns
import matplotlib.pyplot as plt

penguins = sns.load_dataset("penguins")

ax = sns.barplot(data=penguins, x="species", y="body_mass_g", color="steelblue", errorbar="sd")

ax.set_ylim(0, 5500)
ax.set_ylabel("Body Mass (g)")
ax.set_xlabel("Species")

plt.title("Penguin Mass by Species")
plt.show()
```

## Plot 3

The aviation board wants to see passenger growth month-over-month for each year in the 1950s, but the lines are indistinguishable.

```python
import seaborn as sns
import matplotlib.pyplot as plt

flights = sns.load_dataset("flights")
sns.lineplot(data=flights, x="month",  y="passengers",  hue="year", palette="flare" )
plt.title("Monthly Passenger Trends (1949–1960)")
plt.ylabel("Passengers")
plt.xlabel("Month")
plt.legend(title="Year", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
